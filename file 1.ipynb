```python
#sum of digit using function 
def sum_of_digit (num):
    total =0
    while num>0:
        digit=num%10
        total=total + digit
        num=num//10
    return total
print(sum_of_digit(1234))  
print(sum_of_digit(567))   
print(sum_of_digit(9))     #
```

```python
def reverse_number(num):
    reverse=0
    while num>0:
        digit=num%10
        reverse=reverse*10 + digit
        num=num//10
    return reverse
print(reverse_number(1234))  
print(reverse_number(567))   
print(reverse_number(1200))  
print(reverse_number(0))    
```

```python
def is_palindrome(num):
    reverse=0
    original=num
    while num>0:
       digit=num%10
       reverse=reverse*10+digit
       num=num//10
    
    if original==reverse:
     return True
    else:
      return False

print(is_palindrome(121))  
print(is_palindrome(123))  
print(is_palindrome(1331))  
print(is_palindrome(0))     
```

```python
def count_even(nums):
    count=0
    for i in nums:
        if i%2==0:
            count=count+1
    return count
   

print(count_even([1, 2, 3, 4, 5, 6]))  # 3
print(count_even([2, 8, 10]))          # 3
print(count_even([1, 3, 5]))           # 0
print(count_even([]))                  # 0
```

```python
def find_largest(nums):
    largest=nums[0]
    for i in nums:
        if i>largest:
            largest=i
    return largest   
print(find_largest([10, 25, 5, 40, 15]))  # 40
print(find_largest([3, 8, 2]))            # 8
print(find_largest([-5, -2, -10]))        # -2
```

```python
def second(nums):
    largest = nums[0]
    second = None
    for i in nums:
        if i>largest:
           second=largest
           largest=i
        elif i!=largest and (second is None or second<i):
         second=i
    return second 
      
print(second([10, 20, 4, 15]))  # 15
print(second([5, 5,4, 3, 2]))     # 3
print(second([1, 2, 3]))        # 2
```

```python
def  frequency(nums):
    frequency ={}
    for i in nums:
        if i in frequency:
         frequency[i]=frequency[i]+1
        else:
           frequency[i]=1
    return frequency

print(frequency([1, 2, 2, 3, 1, 1]))
# {1: 3, 2: 2, 3: 1}

print(frequency([5, 5, 5]))
# {5: 3}

print(frequency([1, 2, 3]))
# {1: 1, 2: 1, 3: 1}
      
```

```python
def find_duplicates(nums):
    seen=set()
    duplicate=set()
    for i in nums:
       if i in seen:
           duplicate.add(i)
       else:
           seen.add(i)
    return list(duplicate)
print(find_duplicates([1, 2, 3, 2, 4, 1]))
# [1, 2]

print(find_duplicates([5, 5, 5, 2, 2]))
# [5, 2]

print(find_duplicates([1, 2, 3, 4]))
# []
```

```python
#fibonacci series using recursion
def fibonacci(n):
    if n <= 0:
        return []
    elif n == 1:
        return [0]
    elif n == 2:
        return [0, 1]
    else:
        fib_series = fibonacci(n - 1)
        next_value = fib_series[-1] + fib_series[-2]
        fib_series.append(next_value)
        return fib_series
n=int(input("enter the fibonacci series number:")) 
print(fibonacci(n))   
```

```python
# User se number input lena
n = int(input("Enter a number (jaise 5 ya 10): "))

def generate_fibonacci(count):
    if count <= 0:
        return []
    elif count == 1:
        return [0]
    
    fib_list = [0, 1]
    for i in range(2, count):
        next_num = fib_list[-1] + fib_list[-2]
        fib_list.append(next_num)
        
    return fib_list

# Output print karna
result = generate_fibonacci(n)
print(f"Pehle {n} Fibonacci numbers: {result}")

# Agar aapko sirf wahi n-th number chahiye (jaise 5th position wala number):
if n > 0:
    print(f"Is series ka {n}-th number hai: {result[-1]}")
```
